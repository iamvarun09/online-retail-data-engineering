```python
from pyspark.sql import functions as F
#missing value in pyspark
missing_values = df.select([
    F.sum(
        F.when(F.col(c).isNull(), 1).otherwise(0)
    ).alias(c)
    for c in df.columns
])

display(missing_values)
```

```python
#data inspection
from pyspark.sql import functions as F

cancelled_df = df.filter(
    F.col("InvoiceNo").startswith("C")
)

print("Cancelled invoice rows:", cancelled_df.count())

display(cancelled_df.limit(10))
```

```python
bronze_path = "/Volumes/workspace/default/online_retail_data/bronze"

df.write.format("delta").mode("overwrite").save(bronze_path)
```

```python
bronze_df = spark.read.format("delta").load(bronze_path)

print("Bronze row count:", bronze_df.count())

display(bronze_df.limit(5))
```

```python
silver_df = bronze_df.dropDuplicates()

silver_df = silver_df.dropna(
    subset=["InvoiceNo", "StockCode", "Quantity", "UnitPrice"]
)

print("Bronze rows:", bronze_df.count())
print("Silver rows:", silver_df.count())
```

```python
silver_path = "/Volumes/workspace/default/online_retail_data/silver"

silver_df.write.format("delta").mode("overwrite").save(silver_path)
```

```python
silver_check = spark.read.format("delta").load(silver_path)

print("Silver row count:", silver_check.count())

display(silver_check.limit(5))
```

```python
from pyspark.sql import functions as F

silver_df = silver_df.fillna({
    "Description": "Unknown Product"
})
```

```python
silver_df.select(
    F.sum(
        F.when(F.col("Description").isNull(), 1).otherwise(0)
    ).alias("Missing_Description")
).show()
```

```python
silver_df.write.format("delta") \
    .mode("overwrite") \
    .save(silver_path)
```

```python
silver_check = spark.read.format("delta").load(silver_path)

print("Silver row count:", silver_check.count())

display(silver_check.limit(5))
```

```python
silver_df.printSchema()
```

```python
from pyspark.sql import functions as F

silver_df = silver_df.withColumn(
    "InvoiceDate",
    F.to_timestamp("InvoiceDate", "M/d/yyyy H:mm")
)

silver_df = silver_df.withColumn(
    "Quantity",
    F.col("Quantity").cast("int")
)

silver_df = silver_df.withColumn(
    "UnitPrice",
    F.col("UnitPrice").cast("double")
)
```

```python
silver_df.write.format("delta") \
    .mode("overwrite") \
    .save(silver_path)
```

```python
silver_df = silver_df.withColumn(
    "SalesAmount",
    F.col("Quantity") * F.col("UnitPrice")
)

display(
    silver_df.select(
        "InvoiceNo",
        "StockCode",
        "Quantity",
        "UnitPrice",
        "SalesAmount"
    ).limit(10)
)
```

```python
from pyspark.sql import functions as F

silver_df = silver_df.withColumn(
    "SalesAmount",
    F.round(F.col("SalesAmount"), 2)
)

silver_df.write.format("delta") \
    .mode("overwrite") \
    .save(silver_path)

display(silver_df.select(
    "InvoiceNo", "Quantity", "UnitPrice", "SalesAmount"
).limit(10))
```

```python
gold_country_sales = silver_df.groupBy("Country").agg(
    F.round(F.sum("SalesAmount"), 2).alias("TotalSales"),
    F.countDistinct("InvoiceNo").alias("TotalInvoices")
)

display(gold_country_sales.orderBy(F.desc("TotalSales")))
```

```python
gold_path = "/Volumes/workspace/default/online_retail_data/gold_country_sales"

gold_country_sales.write.format("delta") \
    .mode("overwrite") \
    .save(gold_path)
```

```python
gold_check = spark.read.format("delta").load(gold_path)

display(gold_check.orderBy(F.desc("TotalSales")))
```

```python
gold_check.createOrReplaceTempView("gold_country_sales")
```

```python
result = spark.sql("""
    SELECT
        Country,
        TotalSales,
        TotalInvoices
    FROM gold_country_sales
    ORDER BY TotalSales DESC
""")

display(result)
```

```python
# business kpis in gold layer
result = spark.sql("""
    SELECT
        Country,
        ROUND(TotalSales, 2) AS TotalSales,
        TotalInvoices
    FROM gold_country_sales
    ORDER BY TotalSales DESC
    LIMIT 10
""")

display(result)
```

```python
result = spark.sql("""
    SELECT
        Country,
        ROUND(TotalSales, 2) AS TotalSales,
        ROUND(
            TotalSales * 100.0 /
            SUM(TotalSales) OVER (),
            2
        ) AS SalesPercentage
    FROM gold_country_sales
    ORDER BY TotalSales DESC
    LIMIT 10
""")

display(result)
```

```python
silver_df.createOrReplaceTempView("silver_retail")

result = spark.sql("""
    SELECT
        StockCode,
        Description,
        ROUND(SUM(SalesAmount), 2) AS TotalSales,
        SUM(Quantity) AS TotalQuantity
    FROM silver_retail
    GROUP BY StockCode, Description
    ORDER BY TotalSales DESC
    LIMIT 10
""")

display(result)
```

```python
gold_product_sales = silver_df.groupBy(
    "StockCode", "Description"
).agg(
    F.round(F.sum("SalesAmount"), 2).alias("TotalSales"),
    F.sum("Quantity").alias("TotalQuantity")
)
```

```python
gold_product_path = "/Volumes/workspace/default/online_retail_data/gold_product_sales"

gold_product_sales.write.format("delta") \
    .mode("overwrite") \
    .save(gold_product_path)
```

```python
product_check = spark.read.format("delta").load(gold_product_path)

display(
    product_check.orderBy(F.desc("TotalSales")).limit(10)
)
```
